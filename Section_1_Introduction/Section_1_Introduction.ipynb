[NumPy](https://https://numpy.org/devdocs/user/absolute_beginners.html) is Python’s core library for fast numerical computing. Its support of vectorized computations and optimized linear algebra routines makes array and matrix operations both concise and performant.

|||
|---|---|

```python
# Example vectors/matrix inputs
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
A = np.array([[1, 2],[4, 5],[6, 7]])
```

### Basic Commands
| Command | Definition |
|---|---|
| +, -, *, / | Add, subtract, multiply, divide |
| `.T` <br><br> `.reshape()` | Transpose |
| `np.linalg.pinv` | A method from `linalg` collection of NumPy module.<br><br> Performs Gaussian elimination for the system $A\overrightarrow{x}=b$ |
| ** | Exponentiation |
| sqrt() | Square root |


```python
# +, -, *, / (elementwise for arrays)
print("\na + b =", a + b)

# ' (transpose) -- note: for 1D arrays, a.T looks the same
print("\na.T =", a.T)
#.reshape is used to recieve a column vector
print("\na.reshape(-1, 1) =\n", a.reshape(-1, 1))

# matrix reshape example
print("\nA.reshape(-1, 1) =\n", A.reshape(-1, 1))

# \ (MATLAB backslash) — least-squares solution for A x ≈ b
x_pinv = np.linalg.pinv(A) @ b
x_lstsq = np.linalg.lstsq(A, b, rcond=None)[0]  # preferred
print("\nnp.linalg.pinv(A) @ b =", x_pinv)
print("np.linalg.lstsq(A, b)[0] =", x_lstsq)

# . (memberwise operations) -- NumPy uses * for elementwise multiply
print("\na * b =", a * b)

# ^ (exponentiation) — Python uses **
print("\n4**2 =", 4**2)

# sqrt()
print("\nnp.sqrt(16) =", np.sqrt(16).item())

```

### Basic Matrix Operations
| Command (NumPy) | Definition |
|---|---|
| `np.zeros((m, n))` | Creates an $m\times n$ zero matrix. |
| `np.eye(m)` | Creates an $m\times m$ identity matrix. |
| `np.ones((m, n))` | Creates an $m\times n$ matrix of ones. |
| `x, y = A.shape` | Determines the size of the matrix $A$, where $x$ is the number of rows and $y$ is the number of columns in $A$. |
| `A.size` | Determines the total number of elements in $A$. |
| `np.linalg.det(A)` | Determines the determinant of $A$. |


```python
m, n = 1, 2  # for zeros/ones examples

# zeros(m,n)
Z = np.zeros((m, n))
print("np.zeros((1, 2)) =\n", Z)

# eye(m)
I = np.eye(2)
print("\nnp.eye(2) =\n", I)

# ones(m,n)
O = np.ones((m, n))
print("\nnp.ones((1, 2)) =\n", O)

# rows and columns in a
x, y = A.shape
print("\nA =\n", A)
print("\nA.shape =", (x, y), " -> rows =", x, ", cols =", y)

# number of elements in A
print("\nA.size =", A.size)

# det(A) requires square matrix
# Example: det(A.T @ A) (this is 2x2, so determinant is defined)
det_val = np.linalg.det(A.T @ A).item()
print("\ndet(A.T @ A) =", f"{det_val:.4f}")
```

### Function headers and calls
In Python, functions can be defined directly in a notebook using the `def` keyword, and are called with parentheses using the form `y = function_name(x2, x3, x4, x5)`. A function can return multiple values as a tuple, which is typically unpacked as `x, y = function_name(...)`.

The first line of a Python function is the function header: `def function_name(var1, var2, var3):`. A short description is usually placed immediately after the header as a docstring in triple quotes, which can be viewed with `help(function_name)`. Python functions do not use an end statement; the function body ends with indentation end (when the code returns to the previous indentation level).


| Case | Python header / call |
|---|---|
| One output, three inputs | `def function_name(var1, var2, var3):`<br><br>`    return out`<br><br>Call: `out = function_name(var1, var2, var3)` |
| Two outputs, two inputs | `def function_name(var1, var2):`<br><br>`    return out1, out2`<br><br>Call: `out1, out2 = function_name(var1, var2)` |
| No outputs, three inputs | `def function_name(var1, var2, var3):`<br><br>`    ...`<br><br>*(optional* `    return None`*)* <br><br>Call: `function_name(var1, var2, var3)` |
| One output, no inputs | `def function_name():`<br><br>`    return out`<br><br>Call: `out = function_name()` |
| No outputs nor inputs | `def function_name():`<br><br>`    ...`<br><br>Call: `function_name()` |

### Logic operators in Python
Logical operators in Python use keywords such as `and`, `or`, `not`, and comparisons use `==`, `!=`, `<`, `<=`, `>`, `>=`.

| Logic statement | Operator |
|---|---|
| OR | `or` |
| AND | `and` |
| NOT | `not` |
| Equals | `==` |
| Not equal to | `!=` |   
| Less than | `<` |
| Less or equal to | `<=` |
| Greater than | `>` |
| Greater than or equal to | `>=` |

In Python, control flow is defined with keywords like `if`, `for`, `while`. Blocks are determined by indentation (there is no end keyword).
| Command (Python) | Comments |
|---|---|
| `if logic:`<br><br>`    ...`<br><br>`elif other_logic:`<br><br>`    ...`<br><br>`else:`<br><br>`    ...` | If-then statement. Usage of `elif` provides a chain of options. |
| `for i in range(m, n+1, p):`<br><br>`    ...` | For-loop. If the step `p` is omitted, the default increment of `+1` is used. |
| `while logic:`<br><br>`    ...` | While-loop. |
| `break` | Allows to exit a loop prematurely. |
