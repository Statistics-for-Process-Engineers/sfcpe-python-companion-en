
### Linear regression summary (matrix form)

**Notation:**

$A \in \mathbb{R}^{m\times n}$ — design matrix, $y \in \mathbb{R}^{m\times 1}$ — response vector

$\hat{\beta}\in\mathbb{R}^{n\times 1}$ — parameter estimates, $\hat{y}=A\hat{\beta}$ — fitted values

$m$ — number of observations, $n$ — number of parameters, $\alpha$ — significance level

| Quantity | Formula | NumPy / SciPy |
|:---|:---:|:---|
| Parameter estimates | $\displaystyle \hat{\beta} = (A^{T}A)^{-1}A^{T}y$ | `beta = np.linalg.lstsq(A, y, rcond=None)[0]` **(preferred)** <br><br> `beta = np.linalg.inv(A.T @ A) @ (A.T @ y)` |
| Residuals | $\displaystyle \varepsilon = y - A\hat{\beta}$ | `res = y - A @ beta` |
| Model standard deviation | $\displaystyle \hat{\sigma}=\sqrt{\frac{\varepsilon^{T}\varepsilon}{m-n}}$ | `sig_mod = np.sqrt((res.T @ res) / (m - n)).item()` |
| Fisher information matrix | $\displaystyle F=(A^{T}A)^{-1}$ | `F = np.linalg.inv(A.T @ A)` |
| $t$ critical value | $\displaystyle t_{1-\alpha/2,\;m-n}$ | `tcrit = scipy.stats.t.ppf(1 - alpha/2, m - n)` |
| Confidence interval half-width for $\beta_i$ | $\displaystyle \Delta_i=t_{1-\alpha/2,\;m-n}\,\hat{\sigma}\sqrt{F_{ii}}$ | `Delta_i = (tcrit * sig_mod * np.sqrt(F[i, i])).item()` |
| Confidence interval for $\beta_i$ | $\displaystyle \hat{\beta}_i \pm \Delta_i$ | `ci_i = (beta[i,0] - Delta_i, beta[i,0] + Delta_i)` |
| Predicted value at $a_x$ | $\displaystyle \hat{y}=a_x\hat{\beta}$ | `y_hat = (axd @ beta).item()` |
| Mean response confidence interval half-width | $\displaystyle \Delta=t_{1-\alpha/2,\;m-n}\,\hat{\sigma}\sqrt{a_x(A^{T}A)^{-1}a_x^{T}}$ | `Delta = (tcrit * sig_mod * np.sqrt((axd @ F @ axd.T).item()))` |
| Mean response confidence interval | $\displaystyle \hat{y}\pm \Delta$ | `ci_mean = (y_hat - Delta, y_hat + Delta)` |
| Predictive confidence interval half-width | $\displaystyle \Delta=t_{1-\alpha/2,\;m-n}\,\hat{\sigma}\sqrt{1+a_x(A^{T}A)^{-1}a_x^{T}}$ | `Delta = (tcrit * sig_mod * np.sqrt(1 + (axd @ F @ axd.T).item()))` |
| Predictive confidence interval | $\displaystyle \hat{y}\pm \Delta$ | `ci_pred = (y_hat - Delta, y_hat + Delta)` |
| SSR | $\displaystyle SSR=\sum_i(\hat{y}_i-\bar{y})^2$ | `yhat = A @ beta` <br><br> `SSR = np.sum((yhat - y.mean())**2).item()` |
| SSE | $\displaystyle SSE=\sum_i(y_i-\hat{y}_i)^2=\varepsilon^{T}\varepsilon$ | `SSE = (res.T @ res).item()` |
| TSS | $\displaystyle TSS=\sum_i(y_i-\bar{y})^2 = SSR + SSE$ | `TSS = np.sum((y - y.mean())**2).item()` |
| F-statistic | $\displaystyle F=\frac{SSR/k}{SSE/(m-n)}$ | `Fstat = (SSR/k) / (SSE/(m-n))` |
| $R^2$ | $\displaystyle R^2=1-\frac{SSE}{TSS}$ | `R2 = 1 - SSE/TSS` |
| Adjusted $R^2$ | $\displaystyle R^2_{\mathrm{adj}} = 1-\frac{m-1}{m-n}\frac{SSE}{TSS}$ | `R2_adj = 1 - ((m-1)/(m-n))*(SSE/TSS)` |

**Note:** $k$ is the number of model terms (predictors) being tested in the numerator (depends on the hypothesis setup).

### Nonlinear regression

In order to perform nonlinear regression, since nonlinear least-squares problems typically do not have a closed-form solution, in Python a numerical optimizer that could minimize an objective function is required. Once the function minimum is found, the optimal parameter estimates can be used to compute fitted values, residuals, and, with additional derivative information, approximate uncertainty measures such as standard errors and confidence intervals. In Python, the solver tools from `scipy.optimize` module can be used for the task.

It would be easier to define the method that would, given parameters, estimate the sum of residuals sqared

```python
def srs(beta, x, y):
    # beta: vector of parameter values
    # x: regressors for all experiments
    # y: measured values for all experiments
    yhat = model(x, beta)               # predicted values
    return np.sum((y - yhat)**2)        # sum of residuals squared

def model(x, beta):
    # example: yhat = B * x**b
    b_star, B = beta
    return B * x**b_star
```

#### Summary (matrix form)

**Notation:**  
$x \in \mathbb{R}^{m\times 1}$ — regressor values, $y \in \mathbb{R}^{m\times 1}$ — measured values  
$\hat{\beta}\in\mathbb{R}^{n\times 1}$ — estimated parameters, $\hat{y}=g(\hat{\beta},x)$ — fitted values  
$r = y-\hat{y}$ — residual vector, $m$ — number of observations, $n$ — number of parameters, $\alpha$ — significance level  
$J \in \mathbb{R}^{m\times n}$ — Jacobian of residuals, $J_{ij}=\dfrac{\partial r_i}{\partial \beta_j}$

| Quantity | Formula | NumPy / SciPy |
|:---|:---:|:---|
| Predicted values | $\displaystyle \hat{y}=g(\hat{\beta},x)$ | `yhat = model(x, beta)` |
| Residuals | $\displaystyle r=y-\hat{y}$ | `res = y - yhat` |
| Model standard deviation | $\displaystyle \hat{\sigma}=\sqrt{\dfrac{\sum_{i=1}^{m} r_i^2}{m-n}}=\sqrt{\dfrac{r^{T}r}{m-n}}$ | `sig_mod = np.sqrt((res.T @ res) / (m - n)).item()` |
| Fisher information matrix | $\displaystyle F=(J^{T}J)^{-1}$ | `F = np.linalg.inv(J.T @ J)` |
| $t$ critical value | $\displaystyle t_{1-\alpha/2,\;m-n}$ | `tcrit = scipy.stats.t.ppf(1 - alpha/2, m - n)` |
| confidence interval half-width for $\beta_i$ | $\displaystyle \Delta_i=t_{1-\alpha/2,\;m-n}\,\hat{\sigma}\sqrt{F_{ii}}$ | `Delta_i = (tcrit * sig_mod * np.sqrt(F[i, i])).item()` |
| Confidence interval for $\beta_i$ | $\displaystyle \hat{\beta}_i \pm \Delta_i$ | `ci_i = (beta[i,0] - Delta_i, beta[i,0] + Delta_i)` |
| Predicted value at $x_d$ | $\displaystyle \hat{y}_d=g(\hat{\beta},x_d)$ | `y_hat = model(xd, beta)` |
| Mean response confidence interval half-width | $\displaystyle \Delta=t_{1-\alpha/2,\;m-n}\,\hat{\sigma}\sqrt{J_x\,F\,J_x^{T}}$ | `Delta = (tcrit * sig_mod * np.sqrt((Jx @ F @ Jx.T).item())).item()` |
| Mean response CI | $\displaystyle \hat{y}_d \pm \Delta$ | `ci_mean = (y_hat - Delta, y_hat + Delta)` |
| Predictive confidence interval half-width | $\displaystyle \Delta=t_{1-\alpha/2,\;m-n}\,\hat{\sigma}\sqrt{1+J_x\,F\,J_x^{T}}$ | `Delta = (tcrit * sig_mod * np.sqrt((1 + (Jx @ F @ Jx.T).item()))).item()` |
| Predictive confidence interval | $\displaystyle \hat{y}_d \pm \Delta$ | `ci_pred = (y_hat - Delta, y_hat + Delta)` |
| SSR | $\displaystyle SSR=\sum_{i=1}^{m}(\hat{y}_i-\bar{y})^2$ | `SSR = np.sum((yhat - y.mean())**2).item()` |
| SSE | $\displaystyle SSE=\sum_{i=1}^{m}(y_i-\hat{y}_i)^2=r^{T}r$ | `SSE = (res.T @ res).item()` |
| TSS | $\displaystyle TSS=\sum_{i=1}^{m}(y_i-\bar{y})^2 = SSR + SSE$ | `TSS = np.sum((y - y.mean())**2).item()` |
| F-statistic | $\displaystyle F=\dfrac{SSR/k}{SSE/(m-n)}$ | `Fstat = (SSR/k) / (SSE/(m-n))` |
| F-critical | $\displaystyle F_{1-\alpha,\;k,\;m-n}$ | `Fcrit = scipy.stats.f.ppf(1 - alpha, k, m - n)` |
| $R^2$ | $\displaystyle R^2=\frac{SSR}{TSS}=1-\frac{SSE}{TSS}$ | `R2 = 1 - SSE/TSS` |
| Adjusted $R^2$ | $\displaystyle R^2_{\mathrm{adj}} = 1 - \frac{m-1}{m-n}\frac{SSE}{TSS}$ | `R2_adj = 1 - ((m-1)/(m-n))*(SSE/TSS)` |

*Note:* $k$ is the number of model terms (predictors) tested in the numerator (depends on the hypothesis setup).  
*Note:* $J_x$ denotes the Jacobian row vector evaluated at the prediction point $x_d$: $J_x=\left.\dfrac{\partial r}{\partial \beta}\right|_{x=x_d,\ \beta=\hat{\beta}}$.
