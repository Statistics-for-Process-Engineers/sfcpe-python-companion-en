## Section 3.1.1. Python module `scipy.stats` and methods for Statistical Distributions
According to the [SciPy main manual](https://docs.scipy.org/doc/scipy/tutorial/stats.html), the `scipy.stats` module provides unified access to the main theoretical distributions used in mathematical statistics (normal, t-, χ²-, F-distributions, binomial, Poisson, etc.), as well as to the associated distribution functions, densities, quantile functions, and random number generation procedures. This allows us to focus on formulating and interpreting statistical problems without being distracted by low-level implementation details.

The module's interface is uniform: each distribution is represented by an object with a set of standard methods (`pdf`/`pmf`, `cdf`, `ppf`, `rvs`, etc.). Thanks to this, learning how to work with one distribution is almost automatically transferred to the others, which makes further calculations and code more transparent and easy to maintain.

`scipy.stats` is tightly integrated with previously intoduced `NumPy`: all operations are performed on `ndarray` arrays and support vectorization.


All continuous distributions in SciPy use the same parameter scheme:

`loc` — shift (usually the mathematical expectation μ),

`scale` — scale (usually the standard deviation σ, not the variance),

`size` — the shape of the random value array (analogous to dimensions in MATLAB).


#### Normal distribution: `stats.norm`

## Section 3.1.2. Hypothesis Testing


Let

- $\theta$ be the parameter of interest,
- $t_{\text{computed}}$ be the value of the test statistic computed from the sample,
- $t_{\text{crit}}$ be the critical value of the test statistic,
- $\alpha$ be the $\alpha$ - error.


```python
# For this part, the parameters are chosen arbitrarily
# General statistical parameters
alpha = 0.05
# Example test statistic
t_computed = 1.915

# Sample reference distribution:
# Student's t-distribution with 7 degrees of freedom
df = 7


dist = stats.t(df=df)

# Case 1a: two-sided, symmetric distribution
t_crit_case1_sym = dist.ppf(1 - alpha / 2.0)

# Case 1b: two-sided, asymmetric distribution
dist = stats.chi2(df=df)
t_crit_lower_case1 = dist.ppf(alpha / 2.0)
t_crit_upper_case1 = dist.ppf(1 - alpha / 2.0)
```

```python
# Case 2: left-sided test
t_crit_case2 = dist.ppf(alpha)
```

```python
# Case 3: right-sided test
t_crit_case3 = dist.ppf(1 - alpha)
```

```python
# Case 1: two-sided test, symmetric distribution (e.g. t-test)
dist = stats.t(df=df)
tail_prob = dist.cdf(-abs(t_computed))
p_case1 = 2 * tail_prob
reject_case1 = p_case1 <= alpha
```

```python
# Case 2: left-sided test
p_case2 = dist.cdf(t_computed)        # P(T <= t_computed)
reject_case2 = p_case2 <= alpha
```

```python
# Case 3: right-sided test
p_case3 = 1 - dist.cdf(t_computed)    # P(T >= t_computed)
reject_case3 = p_case3 <= alpha
```
