### Section 3.1.1. Python module `scipy.stats` and methods for Statistical Distributions
According to the [SciPy main manual](https://docs.scipy.org/doc/scipy/tutorial/stats.html), the `scipy.stats` module provides unified access to the main theoretical distributions used in mathematical statistics (normal, t-, χ²-, F-distributions, binomial, Poisson, etc.), as well as to the associated distribution functions, densities, quantile functions, and random number generation procedures. This allows us to focus on formulating and interpreting statistical problems without being distracted by low-level implementation details.

The module's interface is uniform: each distribution is represented by an object with a set of standard methods (`pdf`/`pmf`, `cdf`, `ppf`, `rvs`, etc.). Thanks to this, learning how to work with one distribution is almost automatically transferred to the others, which makes further calculations and code more transparent and easy to maintain.

`scipy.stats` is tightly integrated with previously intoduced `NumPy`: all operations are performed on `ndarray` arrays and support vectorization.


All continuous distributions in SciPy use the same parameter scheme:

`loc` — shift (usually the mathematical expectation μ),

`scale` — scale (usually the standard deviation σ, not the variance),

`size` — the shape of the random value array (analogous to dimensions in MATLAB).


#### Normal distribution: `stats.norm`

`stats.norm.ppf(q, loc, scale)`: -- percent point function / inverse cumulative distribution function.

```python
stats.norm.ppf(0.95, loc=0, scale = 1)
```

`stats.norm.pdf(x, loc, scale)` -- distribution density at point `x`

```python
stats.norm.pdf(4,0,1)
```

 `stats.norm.cdf(x, loc, scale)` -- сumulative distribution function at point `x`

```python
stats.norm.cdf(0.5,0,1)
```

 `stats.norm.rvs(loc, scale, size=...)`  -- generation of random variables; for example, `size=(2, 1)` corresponds to a 2×1 matrix.

Important: `scale` is σ, not σ². While MATLAB typically uses variance, Python requires standard deviation.

```python
stats.norm.rvs(loc=0,scale=1,size=(2,1))
```

### Section 3.1.2. Hypothesis Testing


Let

- $\theta$ be the parameter of interest,
- $t_{\text{computed}}$ be the value of the test statistic computed from the sample,
- $t_{\text{crit}}$ be the critical value of the test statistic,
- $\alpha$ be the $\alpha$ - error.

Three basic cases according to the alternative hypothesis $H_1$ can be determined:


#### Case 1: $H_1 : \theta \neq \hat{\theta}$ (two-sided alternative)

**(a) Symmetric distribution** (e.g. Student’s t).

The rejection region is split between the two tails.  
$H_0$ is rejected if

$$
\bigl| t_{\text{computed}} \bigr| \ge t_{\text{crit}},
\qquad
t_{\text{crit}} = \mathrm{inv}(1 - 0.5\alpha).
$$

**(b) Non-symmetric distribution** (e.g. $\chi^2$).

Different lower and upper critical values are used:

$$
t_{\text{crit,lower}} = \mathrm{inv}(0.5\alpha), \qquad
t_{\text{crit,upper}} = \mathrm{inv}(1 - 0.5\alpha),
$$

and $H_0$ is rejected if

$$
t_{\text{computed}} \le t_{\text{crit,lower}}
\quad\text{or}\quad
t_{\text{computed}} \ge t_{\text{crit,upper}}.
$$

#### Case 2: $H_1 : \theta < \hat{\theta}$ (left-sided alternative)

The rejection region lies in the left tail.  
$H_0$ is rejected if

$$
t_{\text{computed}} \le t_{\text{crit}},
\qquad
t_{\text{crit}} = \mathrm{inv}(\alpha).
$$


```python
# Case 2: left-sided test
t_crit_case2 = dist.ppf(alpha)
```

#### Case 3: $H_1 : \theta > \hat{\theta}$ (right-sided alternative)

The rejection region lies in the right tail.  
We reject $H_0$ if

$$
t_{\text{computed}} \ge t_{\text{crit}},
\qquad
t_{\text{crit}} = \mathrm{inv}(1 - \alpha).
$$

```python
# Case 3: right-sided test
t_crit_case3 = dist.ppf(1 - alpha)
```

In certain applications, instead of computing a critical value, the decision is based on the $p$-value of the computed test statistic $t_{\text{computed}}$.
For each case the $p$-value can be expressed in terms of the cumulative
distribution function $\text{cdf}$ of the test statistic under $H_0$.


1. Case 1: $H_1 : \theta \neq \hat{\theta}$ (two-sided)

   a. **Symmetric distribution** (e.g. Student’s):

   The two-sided p-value is obtained by doubling the left-tail probability
   below $-\,|t_{\text{computed}}|$:
   $$
   p = 2\,\text{cdf}\bigl(-|t_{\text{computed}}|\bigr).
   $$
   $H_0$ is rejected if
   $$
   p \le \alpha.
   $$

   This simple formula relies on symmetry and does not apply to
   non-symmetric distributions in the Case 1.

```python
# Case 1: two-sided test, symmetric distribution (e.g. t-test)
dist = stats.t(df=df)
tail_prob = dist.cdf(-abs(t_computed))
p_case1 = 2 * tail_prob
reject_case1 = p_case1 <= alpha
```




2. Case 2: $H_1 : \theta < \hat{\theta}$ (left-sided)

   The rejection region is in the left tail, so the $p$-value is just the
   left-tail probability at the observed statistic:
   $$
   p = \text{cdf}\bigl(t_{\text{computed}}\bigr),
   $$
   and $H_0$ is rejected if
   $$
   p \le \alpha.
   $$



```python
# Case 2: left-sided test
p_case2 = dist.cdf(t_computed)        # P(T <= t_computed)
reject_case2 = p_case2 <= alpha
```


3. Case 3: $H_1 : \theta > \hat{\theta}$ (right-sided)  

   The rejection region is in the right tail, so the p-value is the
   corresponding upper-tail probability:
   $$
   p = 1 - \text{cdf}\bigl(t_{\text{computed}}\bigr),
   $$
   and we reject $H_0$ if
   $$
   p \le \alpha.
   $$

Here $\text{cdf}$ is the cumulative distribution function for the test statistic.

```python
# Case 3: right-sided test
p_case3 = 1 - dist.cdf(t_computed)    # P(T >= t_computed)
reject_case3 = p_case3 <= alpha
```
