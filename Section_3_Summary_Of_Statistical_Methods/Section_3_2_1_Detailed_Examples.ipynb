### Section 3.2.1. Testing a Mean Against a Theoretical Value



```python
X = np.array([2.16, 2.71, 1.09, 0.40, 1.47, 1.13, 1.97])
mu0 = 1.0           # theoretical mean
alpha = 0.05
```

```python
n = len(X)                         # sample size
mu_hat = np.mean(X)                # sample mean
sig = np.std(X, ddof=1)            # sample standard deviation (unbiased)


print(f"Sample size (n):           {n}")
print(f"Sample mean (mu_hat):      {mu_hat:.4f}")
print(f"Sample std. dev. (s):      {s:.4f}")
```

```python
t_computed = (mu_hat - mu0) / (sig / np.sqrt(n))
print(f"Test statistic t_computed: {t_computed:.4f}")
```

For a two-sided test,

```python
df = n - 1
dist = stats.t(df=df)
```

```python
t_crit = dist.ppf(1 - alpha / 2)
print(f"Critical value t_crit (two-sided, alpha = {alpha}): {t_crit:.4f}")
```

```python
reject_critical = abs(t_computed) >= t_crit


print(f"|t_computed| = {abs(t_computed):.4f}")
print(f"t_crit      = {t_crit:.4f}")
print(f"|t_computed| >= t_crit is {reject_critical}")
print("Decision at alpha = 0.05:", "reject H0" if reject_critical else "do not reject H0")
```

For the data sample,

|t_{\text{computed}}| \approx 1.915 < t_{\text{crit}} \approx 2.447.
\$

Therefore

$
|t_{\text{computed}}| \ge t_{\text{crit}}\quad \text{is false},
$

so at a significance level of 5%, the null hypothesis $(H_0)$ **is not rejected**.
The data is consistent with the assumption that the true mean of the population
is equal to $(\mu_0 = 1)$.


The *p*-value corresponding to the test statistic is

$$
p = 2\,P\bigl(T \le -|t_{\text{computed}}|\bigr),
$$

where $(T$) has a Student’s *t*-distribution with $(df = n-1$) degrees of freedom under $(H_0$).
Comparing the *p*-value with the significance level $(\alpha = 0.05)$, the condition

$$
p \le \alpha
$$

is false. Thus, the *p*-value approach leads to the same conclusion as the critical-value approach. We do **not** reject the null hypothesis, and the sample is consistent with a population mean equal to $(\mu_0 = 1)$.


```python
tail_prob = stats.t.cdf(-abs(t_computed), df = n - 1)   # P(T <= −|t|)
p_value = 2 * tail_prob

print(f"tail probability P(T <= -|t|): {tail_prob:.4f}")
print(f"two-sided p-value:             {p_value:.4f}")
print(f"p_value <= alpha ?             {p_value <= alpha}")
```


Comparing the *p*-value with the significance level $(\alpha = 0.05)$, the condition

$$
p \le \alpha
$$

is false. Thus, the *p*-value approach leads to the same conclusion as the critical-value approach. We do **not** reject the null hypothesis, and the sample is consistent with a population mean equal to $(\mu_0 = 1)$.


Finally, to construct a 95% confidence interval for the unknown mean,
a normal population with unknown variance, the confidence interval based on the
Student’s *t*-distribution is

$$
\hat{\mu} \pm t_{1-\alpha/2,\,df}\,\frac{s}{\sqrt{n}},
$$

where $\hat{\mu}$ is the sample mean, $s$ is the unbiased sample standard deviation,
$n$ is the sample size, and $df = n - 1$.

```python
margin = t_crit * sig / np.sqrt(n)
lower = mu_hat - margin
upper = mu_hat + margin


print(f"Margin of error:          {margin:.4f}")
print(f"95% confidence interval: [{lower:.4f}, {upper:.4f}]")
```

```python
t_stat, p_val_builtin = stats.ttest_1samp(X, popmean=mu0)
t_stat, p_val_builtin
```

Since the theoretical mean $\mu_0 = 1$ lies inside this interval, it can be concluded that the confidence interval is compatible with the hypothesis $\mu = \mu_0$.
