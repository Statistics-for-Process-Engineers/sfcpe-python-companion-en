```python
import numpy as np
import scipy.stats as stats
```

### Section 3.2.2. Comparing Two Sample Variances

The task is to compare two drying options, **A** and **B**, using the following product quality data:

- Option A:
  $$
  \{95.6,\ 97.3,\ 95.6,\ 95.4,\ 99.4,\ 97.2,\ 92.2,\ 92.8,\ 94.3,\ 92.6\}
  $$
- Option B:
  $$
  \{89.2,\ 94.2,\ 93.9,\ 93.2,\ 94.7,\ 91.7,\ 93.2,\ 92.4,\ 91.8,\ 91.5\}
  $$

Each option is assumed to produce normally distributed quality measurements.
At significance level $\alpha = 0.05$, it is required to test whether the variance for option B is greater than the variance for option A. If the variance of option B is larger, the new method B should not be selected.

```python
A = np.array([95.6, 97.3, 95.6, 95.4, 99.4, 97.2, 92.2, 92.8, 94.3, 92.6])
B = np.array([89.2, 94.2, 93.9, 93.2, 94.7, 91.7, 93.2, 92.4, 91.8, 91.5])
```

```python
n_A = len(A)
n_B = len(B)

sig_A = np.std(A, ddof=1)   # unbiased std dev for A
sig_B = np.std(B, ddof=1)   # unbiased std dev for B

print(f"n_A = {n_A},  n_B = {n_B}")
print(f"Sample std dev for A (s_A): {sig_A:.4f}")
print(f"Sample std dev for B (s_B): {sig_B:.4f}")
```

The formal hypothesis test for comparing the two variances can be written as

$$
H_0:\ \sigma_B^2 = \sigma_A^2, \qquad
H_1:\ \sigma_B^2 > \sigma_A^2.
$$

 The F-distribution is required and that the test statistic is simply a ratio between the two variances.

```python
F_comp = (sig_B) / (sig_A)

print(f"F_comp = s_B^2 / s_A^2 = {F_comp:.4f}")
```

```python
alpha = 0.05
df1 = n_B - 1
df2 = n_A - 1

dist_F = stats.f(df1, df2)
F_crit = dist_F.ppf(1 - alpha)

print(f"Degrees of freedom: df1 = {df1}, df2 = {df2}")
print(f"Critical value F_crit (alpha = {alpha}): {F_crit:.4f}")
print(f"F_comp >= F_crit ?  {F_comp >= F_crit}")
```


Since

$$
F_{\text{comp}} \ge F_{\text{crit}}
$$

is false, the hypothesis $H_0$ is **not** rejected


The corresponding p-value can be computed in the following way:
$$
p = P(F \ge F_{\text{comp}}),
$$

 Using `scipy.stats.f.cdf` this is computed as



```python
p_value = 1 - dist_F.cdf(F_comp)

print(f"p-value (right-tailed F test): {p_value:.4f}")
print(f"p_value <= alpha ?            {p_value <= alpha}")
```

Comparing with $\alpha = 0.05$,

$$
p \le \alpha
$$

is again false. Thus, the p-value approach leads to the same conclusion as the
critical-value approach: we do **not** reject the null hypothesis. This is a test.
